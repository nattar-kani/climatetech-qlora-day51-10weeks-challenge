Loading the model in 4-bit mode

````python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig

model_id = "Qwen/Qwen2.5-1.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_id)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
)

model = AutoModelForCausalLM.from_pretrained(
    model_id,
    quantization_config=quantization_config,
    device_map="auto",
)

model.config.use_cache = False

print("Model loaded:", model_id)
print("GPU:", torch.cuda.get_device_name(0))
print("4-bit loading:", model.is_loaded_in_4bit)
````

Preparing the training model

````python
def format_example(example):
    messages = [
        {
            "role": "system",
            "content": (
                "You assess sustainability information carefully. "
                "Distinguish potential contributions from reported outcomes. "
                "Do not claim verified impact without supporting evidence. "
                "Follow the requested JSON schema."
            ),
        },
        {
            "role": "user",
            "content": (
                f"{example['instruction']}\n\n"
                f"Supplied information:\n{example['input']}"
            ),
        },
        {
            "role": "assistant",
            "content": example["output"],
        },
    ]

    return {
        "text": tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=False,
        )
    }


formatted = {}

for split in ["train", "validation", "test"]:
    formatted[split] = dataset[split].map(format_example)

print("Train examples:", len(formatted["train"]))
print("Validation examples:", len(formatted["validation"]))
print("Test examples:", len(formatted["test"]))

print("\nFORMATTED EXAMPLE:")
print(formatted["train"][0]["text"][:1500])
````

LoRA Configuration

````python
from peft import LoraConfig, prepare_model_for_kbit_training, get_peft_model

model = prepare_model_for_kbit_training(model)

lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
    target_modules=[
        "q_proj",
        "k_proj",
        "v_proj",
        "o_proj",
        "gate_proj",
        "up_proj",
        "down_proj",
    ],
)

model = get_peft_model(model, lora_config)

model.print_trainable_parameters()
````

Configuring supervised fine-tuning

````python
import torch

for name, param in model.named_parameters():
    if param.requires_grad:
        param.data = param.data.to(torch.float32)

from collections import Counter

print("Trainable parameter dtypes:")
print(Counter(
    str(p.dtype)
    for p in model.parameters()
    if p.requires_grad
))
````

````python
from trl import SFTConfig, SFTTrainer

training_args = SFTConfig(
    output_dir="./climatetech-qlora",
    num_train_epochs=2,
    per_device_train_batch_size=1,
    per_device_eval_batch_size=1,
    gradient_accumulation_steps=8,
    learning_rate=2e-4,
    logging_steps=10,
    eval_strategy="epoch",
    save_strategy="epoch",
    save_total_limit=2,
    fp16=True,
    bf16=False,
    gradient_checkpointing=True,
    max_length=1024,
    packing=False,
    report_to="none",
)

trainer = SFTTrainer(
    model=model,
    args=training_args,
    train_dataset=formatted["train"],
    eval_dataset=formatted["validation"],
    processing_class=tokenizer,
)

````

````python

import json
import torch
from tqdm.auto import tqdm

def generate_prediction(example):
    messages = [
        {
            "role": "system",
            "content": (
                "You assess sustainability information carefully. "
                "Distinguish potential contributions from reported outcomes. "
                "Do not claim verified impact without supporting evidence. "
                "Follow the requested JSON schema."
            ),
        },
        {
            "role": "user",
            "content": (
                f"{example['instruction']}\n\n"
                f"Supplied information:\n{example['input']}"
            ),
        },
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=600,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id,
        )

    new_tokens = outputs[0][inputs["input_ids"].shape[1]:]
    return tokenizer.decode(new_tokens, skip_special_tokens=True)


results = []

for example in tqdm(dataset["test"], desc="Evaluating"):
    pred = generate_prediction(example)
    reference = example["output"]

    try:
        json.loads(pred)
        valid_json = True
    except json.JSONDecodeError:
        valid_json = False

    try:
        exact_match = (
            json.dumps(json.loads(pred), sort_keys=True)
            == json.dumps(json.loads(reference), sort_keys=True)
        )
    except json.JSONDecodeError:
        exact_match = False

    results.append({
        "prediction": pred,
        "reference": reference,
        "valid_json": valid_json,
        "exact_match": exact_match,
    })

valid_count = sum(r["valid_json"] for r in results)
exact_count = sum(r["exact_match"] for r in results)

print(f"\nTest examples: {len(results)}")
print(f"Valid JSON: {valid_count}/100 ({valid_count / len(results):.1%})")
print(f"Exact match: {exact_count}/100 ({exact_count / len(results):.1%})")
````

````python
import json
import re

def extract_numbers(text):
    return set(re.findall(r"\d+(?:,\d{3})*(?:\.\d+)?", text))

claim_number_checks = []
source_type_checks = []

for i, result in enumerate(results):
    pred = json.loads(result["prediction"])
    ref = json.loads(result["reference"])

    pred_evidence = pred.get("evidence_assessment", {})
    ref_evidence = ref.get("evidence_assessment", {})

    reference_claim = ref_evidence.get("impact_claim", "")
    predicted_claim = pred_evidence.get("impact_claim", "")

    reference_numbers = extract_numbers(reference_claim)
    predicted_numbers = extract_numbers(predicted_claim)

    # Only assess number preservation when the reference claim contains numbers.
    if reference_numbers:
        claim_number_checks.append({
            "index": i,
            "reference_numbers": reference_numbers,
            "predicted_numbers": predicted_numbers,
            "preserved": reference_numbers.issubset(predicted_numbers),
        })

    if "source_type" in ref_evidence:
        source_type_checks.append({
            "index": i,
            "reference": ref_evidence["source_type"],
            "predicted": pred_evidence.get("source_type"),
            "correct": (
                ref_evidence["source_type"]
                == pred_evidence.get("source_type")
            ),
        })

print("EXPLICIT CLAIM NUMBER PRESERVATION")
if claim_number_checks:
    preserved = sum(x["preserved"] for x in claim_number_checks)
    print(f"Preserved: {preserved}/{len(claim_number_checks)}")
    print("\nCases where numbers were not preserved:")
    for item in claim_number_checks:
        if not item["preserved"]:
            print(
                f"Index {item['index']}: "
                f"reference={item['reference_numbers']}, "
                f"prediction={item['predicted_numbers']}"
            )
else:
    print("No reference impact claims contained numbers.")

print("\nSOURCE-TYPE AGREEMENT")
if source_type_checks:
    correct = sum(x["correct"] for x in source_type_checks)
    print(f"Agreement: {correct}/{len(source_type_checks)} "
          f"({correct / len(source_type_checks):.1%})")
````

````python
import json
import time
from pathlib import Path
import torch

evaluation_dir = Path("/content/drive/MyDrive/climatetech-qlora/evaluation")
output_file = evaluation_dir / "base_predictions.jsonl"

# Load any predictions already saved.
saved = {}
if output_file.exists():
    with output_file.open("r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                row = json.loads(line)
                saved[row["index"]] = row

print(f"Already saved: {len(saved)}/100 predictions")

system_prompt = (
    "You assess sustainability information carefully. "
    "Distinguish potential contributions from reported outcomes. "
    "Do not claim verified impact without supporting evidence. "
    "Follow the requested JSON schema."
)

start_time = time.time()

with output_file.open("a", encoding="utf-8") as f:
    for i, example in enumerate(dataset["test"]):
        if i in saved:
            continue

        messages = [
            {"role": "system", "content": system_prompt},
            {
                "role": "user",
                "content": (
                    f"{example['instruction']}\n\n"
                    f"Supplied information:\n{example['input']}"
                ),
            },
        ]

        prompt = tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True,
        )
        inputs = tokenizer(prompt, return_tensors="pt").to(base_model.device)

        with torch.no_grad():
            output_ids = base_model.generate(
                **inputs,
                max_new_tokens=600,
                do_sample=False,
                pad_token_id=tokenizer.eos_token_id,
            )

        new_tokens = output_ids[0][inputs["input_ids"].shape[1]:]
        prediction = tokenizer.decode(
            new_tokens, skip_special_tokens=True
        ).strip()

        row = {
            "index": i,
            "reference": example["output"],
            "prediction": prediction,
        }

        f.write(json.dumps(row, ensure_ascii=False) + "\n")
        f.flush()

        if (i + 1) % 10 == 0:
            print(f"Processed through example {i + 1}/100")

print(f"\nCompleted in {(time.time() - start_time) / 60:.1f} minutes.")
print(f"Saved predictions: {output_file}")
````

Summary

````python
import json
import re
from pathlib import Path

evaluation_dir = Path("/content/drive/MyDrive/climatetech-qlora/evaluation")
predictions_file = evaluation_dir / "base_predictions.jsonl"
summary_file = evaluation_dir / "base_evaluation_summary.json"

def parse_json(text):
    """Parse JSON, including responses wrapped in Markdown code fences."""
    text = text.strip()
    text = re.sub(r"^```(?:json)?\s*", "", text, flags=re.IGNORECASE)
    text = re.sub(r"\s*```$", "", text)
    try:
        return json.loads(text)
    except (json.JSONDecodeError, TypeError):
        return None

def canonical_json(obj):
    return json.dumps(obj, sort_keys=True, ensure_ascii=False)

with predictions_file.open("r", encoding="utf-8") as f:
    results = [json.loads(line) for line in f if line.strip()]

results.sort(key=lambda row: row["index"])

valid_json = 0
exact_matches = 0
schema_matches = 0

for row in results:
    prediction = parse_json(row["prediction"])
    reference = parse_json(row["reference"])

    if prediction is not None:
        valid_json += 1

        if isinstance(prediction, dict) and isinstance(reference, dict):
            if set(prediction.keys()) == set(reference.keys()):
                schema_matches += 1

            if canonical_json(prediction) == canonical_json(reference):
                exact_matches += 1

n = len(results)

summary = {
    "model": "Qwen/Qwen2.5-1.5B-Instruct (base, 4-bit)",
    "test_examples": n,
    "valid_json_count": valid_json,
    "valid_json_rate_percent": round(100 * valid_json / n, 2) if n else 0,
    "matching_top_level_schema_count": schema_matches,
    "matching_top_level_schema_rate_percent": round(100 * schema_matches / n, 2) if n else 0,
    "exact_match_count": exact_matches,
    "exact_match_rate_percent": round(100 * exact_matches / n, 2) if n else 0,
}

with summary_file.open("w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2, ensure_ascii=False)

print(json.dumps(summary, indent=2))
print(f"\nSummary saved to: {summary_file}")
````
